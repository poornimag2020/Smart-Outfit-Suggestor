````python

%%writefile app.py

import streamlit as st

# Page configuration
st.set_page_config(
    page_title="Outfit Suggestor",
    page_icon="👕",
    layout="centered"
)

# Title
st.title("👕 Outfit Suggestor")
st.write("Choose the weather and occasion to get a suitable outfit.")

st.divider()

# Weather selection
weather = st.selectbox(
    "🌤️ Select the weather",
    ["Sunny", "Rainy", "Cold"]
)

# Occasion selection
occasion = st.selectbox(
    "🎉 Select the occasion",
    ["Casual", "Formal", "Party"]
)

# Suggest button
if st.button("✨ Suggest Outfit"):

    if weather == "Sunny":

        if occasion == "Casual":
            outfit = "👕 T-shirt + 👖 Jeans + 🕶️ Sunglasses"

        elif occasion == "Formal":
            outfit = "👔 Shirt + 👖 Trousers"

        else:
            outfit = "👗 Stylish Dress + 🕶️ Sunglasses"

    elif weather == "Rainy":

        if occasion == "Casual":
            outfit = "👕 T-shirt + 👖 Jeans + ☂️ Umbrella"

        elif occasion == "Formal":
            outfit = "👔 Formal Shirt + 👖 Trousers + 🧥 Raincoat"

        else:
            outfit = "👗 Party Dress + 🧥 Jacket + ☂️ Umbrella"

    else:  # Cold

        if occasion == "Casual":
            outfit = "🧶 Sweater + 👖 Jeans + 🧥 Jacket"

        elif occasion == "Formal":
            outfit = "👔 Formal Shirt + 👖 Trousers + 🧥 Coat"

        else:
            outfit = "👗 Dress + 🧥 Coat + 👢 Boots"

    # Display result
    st.success("Your suggested outfit is:")

    st.markdown(
        f"### {outfit}"
    )
```
````
